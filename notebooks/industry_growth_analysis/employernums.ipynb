```python
import pandas as pd

ready = pd.read_csv(data_path("AllRoles_nioccs_ready.csv"), encoding="utf-8-sig")

# original numbers
print("Unique employers, all years:", ready["Agency Name"].nunique())
print("Unique employers, 2025 alone:", ready[ready["Cohort Year"] == 2025]["Agency Name"].nunique())

top_repeat = ready["Agency Name"].value_counts()
print("Employers with 5+ postings:", (top_repeat >= 5).sum())

# --- does 5+ postings mean repeat across cycles, or just volume in one cycle? ---
cycles_per_employer = ready.groupby("Agency Name")["Cohort Year"].nunique()
posts_per_employer = ready["Agency Name"].value_counts()

five_plus = posts_per_employer[posts_per_employer >= 5].index
five_plus_cycles = cycles_per_employer.loc[five_plus]

print("\nOf the 5+ posting employers:")
print("  In just 1 cycle:      ", (five_plus_cycles == 1).sum())
print("  Across 2+ cycles:     ", (five_plus_cycles >= 2).sum())
print("  Across 3+ cycles:     ", (five_plus_cycles >= 3).sum())

# a cleaner "true repeat partner" definition, if you want one:
true_repeat_partners = (cycles_per_employer >= 2).sum()
print("\nEmployers active in 2+ distinct cycles/years:", true_repeat_partners)
```

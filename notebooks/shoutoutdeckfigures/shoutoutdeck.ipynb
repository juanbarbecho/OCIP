```python
import pandas as pd

# --- load & normalize old file ---
old = pd.read_csv("Historical SF__CCL Payroll Data-Grid view copy.csv")
old = old.rename(columns={"EMPL ID": "EMPLID", "Program": "Program", "Total Amount Paid": "Total Paid"})
old["Total Paid"] = old["Total Paid"].replace(r"[\$,]", "", regex=True).astype(float)
old = old[["EMPLID", "Program", "Total Paid"]]

# --- load & normalize new file ---
new = pd.read_csv("NEW Students Paid-Grid view.csv")
new = new.rename(columns={"EMPLID": "EMPLID", "Program + Semester": "Program", "Total Pay (from payroll data)": "Total Paid"})
new["Total Paid"] = new["Total Paid"].replace(r"[\$,]", "", regex=True).astype(float)
new = new[["EMPLID", "Program", "Total Paid"]]

# --- combine ---
combined = pd.concat([old, new], ignore_index=True)

print("Total students served:", combined["EMPLID"].nunique())
print("Total $ spent:        ${:,.2f}".format(combined["Total Paid"].sum()))

print("\nBy program:")
print(combined.groupby("Program").agg(
    students=("EMPLID", "nunique"),
    total_paid=("Total Paid", "sum")
).sort_index())
```
